#Método de bisección

```
import math
import matplotlib.pyplot as plt

# 1.-Definimos las funciones solicitadas
def funcion_a(x):
    return (x**3) + 4*(x**2) - 10.0

def funcion_b(x):
    return math.exp(3*x) - 4.0

def funcion_c(x):
    return -0.4*(x**2) + 2.3*x + 2.2

def funcion_d(x):
    return (x**4) + 3*(x**2) - 2.0


# 2.-Método de bisección
def ejecutar_biseccion(funcion_evaluar, Limite_Inferior, Limite_Superior, Error_Esperado, Nombre_Inciso):
    print(f"\n--------------------------------------------------")
    print(f"Iniciando Bisección: {Nombre_Inciso}")
    print(f"--------------------------------------------------")

    Error_Actual = 100.0  # Valor inicial forzado para entrar al loop
    Punto_Medio_Pasado = 0.0
    Iteracion = 1

    # Listas para guardar la información que irá a las gráficas
    Historial_Iteraciones = []
    Historial_Errores = []

    while Error_Actual > Error_Esperado:
        # Cortamos el espacio de búsqueda a la mitad
        Punto_Medio = (Limite_Inferior + Limite_Superior) / 2.0

        # Comparamos este nuevo punto con el de la iteración pasada (Error Relativo)
        if Iteracion > 1:
            Error_Actual = abs((Punto_Medio - Punto_Medio_Pasado) / Punto_Medio)

        # Guardamos datos para la gráfica (en iteración 1 el error es 100%, evitamos graficar ese pico)
        if Iteracion > 1:
            Historial_Iteraciones.append(Iteracion)
            Historial_Errores.append(Error_Actual)

        print(f"Iteracion {Iteracion}: Raíz calculada = {Punto_Medio:.6f} | Error = {Error_Actual:.6f}")

        # Evaluación de la función en los puntos clave
        Funcion_Evaluada_Inferior = funcion_evaluar(Limite_Inferior)
        Funcion_Evaluada_Medio = funcion_evaluar(Punto_Medio)

        # Eliminación de la mitad inútil del intervalo
        if (Funcion_Evaluada_Inferior * Funcion_Evaluada_Medio) < 0:
            Limite_Superior = Punto_Medio    # La raíz está en la izquierda
        elif (Funcion_Evaluada_Inferior * Funcion_Evaluada_Medio) > 0:
            Limite_Inferior = Punto_Medio    # La raíz está en la derecha
        else:
            Error_Actual = 0.0               # Le dimos a la raíz exacta por accidente

        Punto_Medio_Pasado = Punto_Medio
        Iteracion = Iteracion + 1

        # Digamos que es un "freno de emergencia" por si la función no pasa por el cero en ese intervalo
        if Iteracion > 100:
            print("Alerta: Se superaron las 100 iteraciones. Abortando ciclo.")
            break

    print(f"\n--> RAÍZ FINAL APROXIMADA PARA {Nombre_Inciso}: {Punto_Medio:.6f}")

    # ==========================================
    # 3.-Gráficas

    plt.figure(figsize=(6, 4))
    plt.plot(Historial_Iteraciones, Historial_Errores, marker='o', linestyle='-', color='b')
    plt.title(f"Caída del Error - Bisección ({Nombre_Inciso})")
    plt.xlabel("Número de Iteración")
    plt.ylabel("Error Aproximado (Ea)")
    plt.grid(True)
    plt.show()


# 4. Ejecución edn secuencia de los incisos

# Inciso a) Intervalo [1, 2], Ea = 0.0001
ejecutar_biseccion(funcion_a, 1.0, 2.0, 0.0001, "Inciso A")

# Inciso b) Intervalo [0, 1], Ea = 0.01
ejecutar_biseccion(funcion_b, 0.0, 1.0, 0.01, "Inciso B")

# Inciso c) Intervalo [5, 8], Ea = 3% (0.03 decimal)
ejecutar_biseccion(funcion_c, 5.0, 8.0, 0.03, "Inciso C")

# Inciso d) Intervalo [0, 3], Ea = 0.0001
ejecutar_biseccion(funcion_d, 0.0, 3.0, 0.0001, "Inciso D")
```
